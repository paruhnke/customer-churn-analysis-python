# Customer Churn Analysis Using Python

This notebook analyzes gym membership data to identify factors associated with customer churn and retention.

## 1. Import Libraries

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

IMAGES_DIR = Path("../images")
IMAGES_DIR.mkdir(parents=True, exist_ok=True)
```

## 2. Load Dataset

```
df = pd.read_csv("../data/gym_churn.csv")
df.head()
```

## 3. Initial Data Review

```
print(df.shape)
df.info()
```

```
df.describe()
```

## 4. Data Cleaning and Feature Engineering

The churn column is converted into a readable retention label for visualization. Gender is also mapped into readable labels.

```
df["retention"] = df["Churn"].map({0: "Retained", 1: "Left"})
df["gender_label"] = df["gender"].map({0: "Male", 1: "Female"})

# Check for missing values
df.isnull().sum()
```

## 5. Overall Churn Rate

```
churn_rate = df["Churn"].mean() * 100
print(f"Overall churn rate: {churn_rate:.2f}%")
```

## 6. Churn by Gender

```
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="gender_label", hue="retention")
plt.title("Retention by Gender")
plt.xlabel("Gender")
plt.ylabel("Member Count")
plt.tight_layout()
plt.savefig(IMAGES_DIR / "retention_by_gender.png", bbox_inches="tight")
plt.show()

(df.groupby("gender_label")["Churn"].mean() * 100).round(2)
```

## 7. Churn by Location Proximity

```
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Near_Location", hue="retention")
plt.title("Retention by Location Proximity")
plt.xlabel("Near Location")
plt.ylabel("Member Count")
plt.xticks([0, 1], ["No", "Yes"])
plt.tight_layout()
plt.savefig(IMAGES_DIR / "retention_by_location.png", bbox_inches="tight")
plt.show()

(df.groupby("Near_Location")["Churn"].mean() * 100).round(2)
```

## 8. Churn by Employer Partnership

```
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Partner", hue="retention")
plt.title("Retention by Employer Partnership")
plt.xlabel("Partner Company")
plt.ylabel("Member Count")
plt.xticks([0, 1], ["No", "Yes"])
plt.tight_layout()
plt.savefig(IMAGES_DIR / "retention_by_partner_company.png", bbox_inches="tight")
plt.show()
```

## 9. Churn by Referral Program

```
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Promo_friends", hue="retention")
plt.title("Retention by Referral Program")
plt.xlabel("Referral Program")
plt.ylabel("Member Count")
plt.xticks([0, 1], ["No", "Yes"])
plt.tight_layout()
plt.savefig(IMAGES_DIR / "retention_by_referral.png", bbox_inches="tight")
plt.show()
```

## 10. Churn by Age

```
churn_by_age = (df.groupby("Age")["Churn"].mean() * 100).reset_index()

plt.figure(figsize=(10, 5))
sns.lineplot(data=churn_by_age, x="Age", y="Churn", marker="o")
plt.title("Churn Rate by Age")
plt.xlabel("Age")
plt.ylabel("Churn Rate (%)")
plt.tight_layout()
plt.savefig(IMAGES_DIR / "churn_by_age.png", bbox_inches="tight")
plt.show()

churn_by_age.round(2)
```

## 11. Key Findings

- Overall churn rate was approximately 26.5%.
- Members who lived or worked near the gym had a lower churn rate than members who were not nearby.
- Younger members had higher churn rates, while churn generally decreased as age increased.
- Gender did not show a meaningful difference in churn rate.
